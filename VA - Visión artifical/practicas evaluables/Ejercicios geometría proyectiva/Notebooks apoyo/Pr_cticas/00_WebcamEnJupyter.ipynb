# Acceso a la webcam desde Jupyter

Este ejemplo captura una imagen con la webcam conectada al equipo donde se ejecuta el kernel de Jupyter. Usa únicamente OpenCV y Matplotlib, incluidos en `requirements-base.txt`.

En Linux, la cámara debe estar disponible como un dispositivo local, normalmente `/dev/video0`. En WSL, una máquina remota o un contenedor puede ser necesario configurar el acceso al dispositivo.

```python
import cv2
import matplotlib.pyplot as plt

camera = cv2.VideoCapture(0)

try:
    if not camera.isOpened():
        raise RuntimeError("No se ha podido abrir la webcam.")

    ok, frame_bgr = camera.read()
    if not ok:
        raise RuntimeError("No se ha podido capturar una imagen.")
finally:
    camera.release()

frame_rgb = cv2.cvtColor(frame_bgr, cv2.COLOR_BGR2RGB)
plt.imshow(frame_rgb)
plt.axis("off");
```
